```python
sample="The Tokenizer is a necessary and pervasive component of Large Language Models (LLMs), where it translates between strings and tokens (text chunks). Tokenizers are a completely separate stage of the LLM pipeline: they have their own training sets, training algorithms (Byte Pair Encoding), and after training implement two fundamental functions: encode() from strings to tokens, and decode() back from tokens to strings. In this lecture we build from scratch the Tokenizer used in the GPT series from OpenAI. In the process, we will see that a lot of weird behaviors and problems of LLMs actually trace back to tokenization. We'll go through a number of these issues, discuss why tokenization is at fault, and why someone out there ideally finds a way to delete this stage entirely."
tokens=sample.encode("utf-8")
tokens=list(map(int,tokens))
print(tokens)
print("Length of raw text",len(sample))
print("Length of tokens",len(tokens))
```

```python
def get_stats(ids):
  counts={}
  for pair in zip(ids, ids[1:]):
    counts[pair]=counts.get(pair,0)+1
  return counts
stats=get_stats(tokens)
print(sorted(((v,k) for k,v in stats.items()),reverse=True))
```

```python
top_pair = max(stats, key=stats.get)
top_pair
```

```python
def merge(ids, pair, idx):
  # in the list of ints (ids), replace all consecutive occurences of pair with the new token idx
  newids = []
  i = 0
  while i < len(ids):
    # if we are not at the very last position AND the pair matches, replace it
    if i < len(ids) - 1 and ids[i] == pair[0] and ids[i+1] == pair[1]:
      newids.append(idx)
      i += 2
    else:
      newids.append(ids[i])
      i += 1
  return newids

print(merge([5, 6, 6, 7, 9, 1], (6, 7), 99))

tokens2 = merge(tokens, top_pair, 256)
print(tokens2)
print("length:", len(tokens2))
```

have to merge from top to bottom i.e. here from 257 to 276 as the later merges use the top merges eg: (97,258),276

```python
# longer we go down in the get_stats larger will be the vocab and shorter will be the sequence. We have to find a sweet spot between the two!

# decide a vocab size(hyper parameter)
vocab_size=276
num_merges=vocab_size-256
ids=list(tokens) # copying so we dont destroy the orignal list
merges={}
for i in range(num_merges):
    stats = get_stats(ids)
    pair=max(stats,key=stats.get)
    idx=256+i
    print(f"merging {pair} into a new token {idx}")
    ids=merge(ids,pair,idx) # replacing the co-occurance of th pair with idx

    merges[pair]=idx
```

```python
from copy import replace
vocab={idx:bytes([idx]) for idx in range(256)}
#vocab={97:a,65:A}, so when we call vocab([97])-->a
for (p0,p1), idx in merges.items():
  #from merges say {(97,65),257} so vocab([257])=vocab([97])+vocab([65]) or vocab{257:aA}
  vocab[idx]=vocab[p0]+vocab[p1]

def decode(ids):
  tokens=b"".join(vocab[idx] for idx in ids)
  text=tokens.decode("utf-8",errors="replace")
  return text
print(decode([258, 97]))
```

```python
def encode(text):
  # given a string, return list of integers (the tokens)
  tokens = list(text.encode("utf-8"))
  while len(tokens) >= 2:
    stats = get_stats(tokens)
    #MIN is looking at the key of stats i.e pair, so trying to find if min pair in the stats(tokens) matches the merges pair
    pair = min(stats, key=lambda p: merges.get(p, float("inf")))
    if pair not in merges:
      break # nothing else can be merged
    idx = merges[pair]
    tokens = merge(tokens, pair, idx)
  return tokens

print(encode(""))
```
